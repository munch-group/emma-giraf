# Introduktion

## Motivation

- Bavianer (*P. Papio*) som modelsystem for kompleks evolutionær historie
- Seks arter: *P. ursinus, P. kindae, P. cynocephalus, P. anubis, P. hamadryas, P. papio*

- Spørgsmål: $N_e$, migrationsrater og split-tidspunkter

## Populationsgenetisk baggrund

- Wright-Fisher- og Moran-modellen som fundament
- Kingmans koalescent: baglæns i tid, MRCA, koalescensrate
- Effektiv populationsstørrelse $N_e$ og dens biologiske fortolkning
- Site Frequency Spectrum: hvad det måler og hvorfor det er informativt
- Wattersons estimator og nukleotiddiversitet $\pi$
- Tajima’s $D$ som test for neutral evolution
- Struktureret koalescent: migration og populationsopdeling
- Two-island modellen og IM-modellen som modeller
- Neutral evolution som modelantagelse og dens begrænsninger

## Eksisterende metoder og begrænsninger

- PSMC og SMC++: styrker og begrænsninger ved enkelt-populations rekonstruktion
- STRUCTURE og ADMIXTURE: populationsstruktur uden koalescentbasis
- fastsimcoal2 og momi2: SFS-baseret inferens via simulering
- Fælles begrænsning: ingen eksakt analytisk beregning uden Monte Carlo
- Manglende fuld Bayesiansk usikkerhedskvantificering i eksisterende metoder

## Fase-type fordelinger

## Modelantagelser

- Neutral evolution: ingen selektion, ingen rekombination i modellen
- Konstant $N_e$ inden for epoker, virkeligheden er mere gradvis

## Datamæssige begrænsninger

- Neutral evolutio, ingen selektion inkluderet, men Tajima’s D antyder mulig afvigelse
- $n=4$ subsampling: mister information fra de resterende individer
- Kun kromosom 20: er det repræsentativt for hele genomet
- Hypergeometrisk subsampling introducerer stikprøveusikkerhed
- Mutationsraten $\mu=0.9 \times 10^{-8}$ er et estimat med usikkerhed på $\pm 30 \%$
- SVGD approksimerer posterioren ikke eksakt som MCMC
- Epoch-switch som Markov-overgang er en approksimation til deterministisk split
- Symmetrisk migration i to-ø modellen muligvis ikke biologisk korrekt
- Konstant $N_e$ inden for epoker virkeligheden er mere gradvis

### Metodologiske begrænsninger

- SVGD approksimerer posterioren ikke eksakt som MCMC
- Variance collapse ved for høj learning rate
- Epoch-switch i IM-modellen er svær at estimere (identifikabilitetsproblem)

## Hvad kunne man gøre bedre eller anderledes?

- Anvende metoden på andre primatsystemer eller andre taxa med kompleks demografisk historie
- Inkludere asymmetrisk migration på bavian-data
- Sammenligne systematisk med fastsimcoal2 eller momi2 på de samme data

# Konklusion

## Metodisk

- Koalescentens fase-type formulering i phasic reproducerer alle analytiske resultater eksakt
- SVGD med Adam(0.25) og DataPrior/HalfCauchyPrior giver pålidelige estimater under bavian-realistiske betingelser
- To-ø modellen er identificerbar for $\theta_1 >0.5$ og N_OBS $\leq 5000$
- IM-modellens split-tidspunkt estimeres med epoch_switch_rate $>$ 10
- Joint probability-inferens forbedrer estimering ved lavt N_OBS og lav $\theta$
- Multi-feature SFS giver gevinst ved $\theta > 8$ og N_OBS $> 3000$

## Empirisk konklusion

- N_e per population med usikkerhed
- Migrationsrate mellem P. cynocephalus og P. kindae
- Split-tidspunkt $t_s$ fra IM-modellen i år
- Demografisk rekonstruktion: N_e over tid for tre populationer
- Sammenfatning af biologisk fortolkning

## Perspektiv

- Fase-type fordelinger kombineret med SVGD udgør en effektiv og præcis ramme for demografisk inferens
- Metoden er direkte anvendelig på andre systemer med kompleks populationsstruktur
- Fremtidigt arbejde: asymmetrisk migration, multi-locus ARG, selektion
